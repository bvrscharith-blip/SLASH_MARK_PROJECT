```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread(r'C:\Users\bvrsc\Documents\indoor.jpg.png')

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image)
plt.title("Indoor Obstacle Environment")
plt.show()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)

edges = cv2.Canny(gray, 100, 200)

plt.imshow(edges, cmap='gray')
plt.title("Obstacle Detection")
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original Indoor Environment")

plt.subplot(1,2,2)
plt.imshow(edges, cmap='gray')
plt.title("Detected Obstacles")

plt.show()
```
