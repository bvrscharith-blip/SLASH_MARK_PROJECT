```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
documents = [
    "Machine learning is very useful in data science",
    "Machine learning is useful for data science",
    "Dogs and cats are common pets",
    "Artificial intelligence is changing the world"
]

print(documents)
```

```python
vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print(tfidf_matrix)
```

```python
similarity_matrix = cosine_similarity(tfidf_matrix)

print(similarity_matrix)
```

```python
print("PLAGIARISM CHECKER RESULTS\n")

for i in range(len(documents)):
    for j in range(i + 1, len(documents)):

        similarity_score = similarity_matrix[i][j] * 100

        print(f"Document {i+1} vs Document {j+1}")
        print(f"Similarity: {similarity_score:.2f}%")

        if similarity_score > 70:
            print("Potential Plagiarism Detected")
        else:
            print("No Significant Plagiarism")

        print("-"*40)
```

```python
df = pd.DataFrame(similarity_matrix)

df.to_csv("plagiarism_report.csv", index=False)

print("Report Saved Successfully")
```
